```python
import pandas as pd
```

```python
# read Excel file
pd.read_excel("persons.xlsx")
```

```python
# read csv file
pd.read_csv("persons.csv")
```

```python
# read json
pd.read_json("persons.json")
```

```python
# read xml
pd.read_xml("persons.xml")
```

```python
# Excel to json
data = pd.read_excel("persons.xlsx")
data.to_json("ex_to_json.json",orient = "records")
```

```python
# json to csv & Excel
data0 = pd.read_json("persons.json")
data0.to_csv("json_to_csv.csv")
data0.to_excel("json_to_excel.xlsx")
```

```python
# changes in data
data = pd.read_excel("persons.xlsx")
data["life-days"] = data["age"] *365
data.to_csv("ex_to_csv.csv")
data
```
